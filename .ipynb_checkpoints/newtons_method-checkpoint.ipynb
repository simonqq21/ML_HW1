# Import Libraries

```python
import numpy as np 
import matplotlib.pyplot as plt
import argparse
import csv
```

# Define input filename and parameters

```python
filename = "input.txt"
lambda_val = 10000
num_polynomial_bases = 3
```

# Read the input file and store input data points into a list 

```python
datapoints = []
with open(filename, 'r', encoding='utf-8') as file:
    csv_reader = csv.reader(file)
    for row in csv_reader:
        datapoints.append([float(i) for i in row])
num_datapoints = len(datapoints)
print(f"number of datapoints: {num_datapoints}\n")

datapoints_arr = np.zeros((num_datapoints, 2), dtype=np.float64)

i = 0
for datapoint in datapoints:
    datapoints_arr[i,0] = datapoint[0]
    datapoints_arr[i,1] = datapoint[1]
    i+=1;

print(datapoints_arr)
```

# Steepest Descent

## Create A and b

```python
b = np.zeros((num_datapoints), dtype=np.float64)
b = datapoints_arr[:,1]
    
A = np.zeros((num_datapoints, num_polynomial_bases), dtype=np.float64)
for basis in range(num_polynomial_bases):
    A[:,basis] = datapoints_arr[:,0] ** basis
```

## Define learning rate

```python
learning_rate = 0.0001
```

## Initialize weights, biases

```python
x = np.zeros(shape=num_polynomial_bases, dtype=np.float32)
# x = np.array([-0.2314002,  4.90619  ,  3.0238533])
```

## Define error function

```python
def compute_error(A, x, b):
    Ax = np.dot(A, x)
    error = np.sum(np.abs(Ax - b) ** 2)
    return error
```

## Get the LSE gradient

```python
A_t = np.transpose(A)
```

```python
learning_rate = 1e-4
```

```python
for i in range(500):
    x_t = np.transpose(x)
    grad = 2 * (np.dot(x_t, np.dot(A_t, A))-np.dot(A_t, b))
    # print(x)
    # print(grad)
    e = np.zeros(shape=x.shape)
    e[np.argmax(np.abs(grad))] = 1
    # print(e)
    grad = grad * e
    error = compute_error(A, x, b)
    print(f"Total error: {error}")
    x = x - learning_rate * grad
    
```
